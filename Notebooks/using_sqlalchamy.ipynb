### Importing the data to SQL

```python
from sqlalchemy import create_engine
import psycopg2
from pathlib import Path
import pandas as pd
```

```python
engine = create_engine("postgresql+psycopg2://postgres:****@localhost:5432/Ecommerce")
```

```python
for path in Path(r"C:\Users\jatar\Documents\SQL_project\brazilian_ecomm").glob("*.csv"):
    df = pd.read_csv(path)
    df.to_sql(path.stem, engine, index = False)
    print("done", path.name)
```
