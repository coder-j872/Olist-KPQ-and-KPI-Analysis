```python
import psycopg2 as psy
```

```python
hostname = 'localhost'
database = 'Ecommerce'
username = 'postgres'
pwd = 'ram@123'
port_id = 5432

conn = None
cur = None
try :
    
    conn = psy.connect(
        host = hostname,
        dbname = database,
        user = username,
        password = pwd,
        port = port_id)
    
    cur = conn.cursor()
    

except :
    print("error")
finally:
    if cur is not None:
        cur.close()
    if conn is not None:
        conn.close()
```
